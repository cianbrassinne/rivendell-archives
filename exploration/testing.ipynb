# Import libraries

```python
from pathlib import Path

import ebooklib
from ebooklib import epub
from bs4 import BeautifulSoup
```

# Import data

```python
EPUB_PATH = Path("../data/hobbit.epub")
assert EPUB_PATH.exists(), f"EPUB not found: {EPUB_PATH}"

book = epub.read_epub(str(EPUB_PATH))
documents = book.get_items_of_type(ebooklib.ITEM_DOCUMENT)

print(f"Number of XHTML documents: {len(list(documents))}")
print(f"Title: {book.title}")
print(f"Author: {book.get_metadata('DC', 'creator')}")
```

# Testing

## Check a few pages

```python
for i, item in enumerate(book.get_items_of_type(ebooklib.ITEM_DOCUMENT)):
    soup = BeautifulSoup(item.get_content(), "html.parser")
    text = soup.get_text(separator="\n", strip=True)

    print(f"\n--- Document {i} ---")
    print(f"Name: {item.get_name()}")
    print(f"Characters: {len(text)}")
    print(text[:500])

    if i >= 4:
        break
```
